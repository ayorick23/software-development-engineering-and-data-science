## 1. Bucles e Iteraciones

Los bucles permiten ejecutar un bloque de instrucciones de manera repetitiva. Python provee dos estructuras principales: **`while`** y **`for`**.

### Bucle `while`
Ejecuta un bloque de código **mientras una condición lógica se evalúe como verdadera (`True`)**.

```python
# Inicialización de variable de control
i = 1
while i < 5:
    print(f'El valor actual de i es: {i}')
    i += 1  # Incremento para evitar un bucle infinito
```

>&#9888;&#65039; **Precaución con los bucles infinitos:**<br>
>Si olvidamos actualizar la variable de control (por ejemplo, omitir `i += 1`), la condición nunca será falsa y el entorno se bloqueará (mostrando `In [*]` de forma permanente en Jupyter). Para detenerlo, ve al menú **Kernel -> Restart Kernel**.

Opcionalmente, `while` admite una cláusula `else` que se ejecuta al finalizar la iteración de manera normal:

```python
i = 1
while i < 5:
    print(f'El valor actual de i es: {i}')
    i += 1
else:
    print('Bucle finalizado exitosamente')
```

### Bucle `for`
El bucle `for` itera directamente sobre los elementos de cualquier secuencia o colección (*strings*, listas, tuplas, diccionarios o generadores).

```python
valores = [1, 2, 3, 4, 5]

# Iteración directa sobre cada elemento
for elemento in valores:
    print(elemento)
```

```python
# Operación matemática dentro de la iteración
for num in valores:
    print(num ** 2)
```

### La función generadora `range()`

En lugar de crear manualmente secuencias grandes en memoria, `range(inicio, fin, paso)` genera números enteros de forma eficiente:

* **1 argumento `range(stop)`:** Genera enteros desde `0` hasta `stop - 1`.
* **2 argumentos `range(start, stop)`:** Inicia en `start` hasta `stop - 1`.
* **3 argumentos `range(start, stop, step)`:** Especifica el incremento o salto.

```python
range(5)
```

```python
# Conversión de objeto range a lista visible
list(range(5))
```

```python
# Iteración con inicio y fin definidos [3, 4]
for i in range(3, 5):
    print(i)
```

```python
# Iteración con paso definido (números impares entre 1 y 9)
for i in range(1, 10, 2):
    print(i)
```

## 2. Comprensión de Listas (*List Comprehension*)

Es una construcción sintáctica concisa y optimizada para transformar o construir listas a partir de iterables en una sola línea.

**Sintaxis base:** `[expresion for elemento in iterable]`

```python
datos = [2, 3, 4, 5]

# Enfoque tradicional con bucle for y .append():
cuadrados_tradicional = []
for num in datos:
    cuadrados_tradicional.append(num ** 2)
    print(cuadrados_tradicional)
print('Enfoque tradicional:', cuadrados_tradicional)

```

```python
# Enfoque idiomático con List Comprehension:
cuadrados_comprension = [num ** 2 for num in datos]
print('List Comprehension:', cuadrados_comprension)
```

```python
# Comprensión combinada con range()
[x ** 2 for x in range(2, 10)]
```

## 3. Funciones Personalizadas (`def`)

Las funciones encapsulan bloques de lógica reutilizables, reduciendo redundancias y permitiendo la modularización del código.

* Se definen con la palabra clave **`def`**.
* Pueden recibir **parámetros** y devolver valores explícitos con **`return`**.
* Incluyen cadenas de documentación (**Docstrings**) delimitadas por tres comillas dobles `"""`.

```python
def calcular_potencia(base, exponente=2):
    """
    Calcula la potencia de un número.
    
    Parámetros:
    base (float/int): Número base.
    exponente (int): Exponente al que se eleva la base (por defecto 2).
    
    Retorna:
    float/int: Resultado de la exponenciación.
    """
    return base ** exponente
```

```python
# Llamada usando el valor por defecto (elevar al cuadrado)
resultado_1 = calcular_potencia(4)
print('Resultado por defecto (4^2):', resultado_1)

# Llamada especificando argumentos personalizados
resultado_2 = calcular_potencia(2, 3)
print('Resultado personalizado (2^3):', resultado_2)
```

## 4. Expresiones Lambda (Funciones Anónimas)

Una expresión `lambda` permite crear funciones anónimas y desechables de una sola línea.

**Sintaxis:** `lambda argumentos: expresion`

```python
# Función estándar:
# def duplicar(x): return x * 2

# Equivalente como función Lambda:
duplicar = lambda x: x * 2
duplicar(5)
```

## 5. Funciones de Orden Superior: `map()` y `filter()`

### Función `map()`
Aplica una función a **cada uno de los elementos** de un iterable sin necesidad de escribir un bucle explícito.

```python
valores = [1, 2, 3, 4, 5]

# Aplicar una función lambda a toda la lista con map()
cuadrados_map = list(map(lambda n: n ** 3, valores))
cuadrados_map
```

```python
# map() operando sobre múltiples listas en simultáneo
lista_a = [1, 2, 3, 4]
lista_b = [10, 20, 30, 40]
lista_c = [100, 200, 300, 400]

suma_listas = list(map(lambda x, y, z: x + y + z, lista_a, lista_b, lista_c))
suma_listas
```

### Función `filter()`
Filtra los elementos de un iterable evaluando una función que retorna un booleano (`True` o `False`). **Solo los elementos evaluados como `True` formarán parte del resultado**.

```python
# Filtrar números pares en un rango del 1 al 10
pares = list(filter(lambda x: x % 2 == 0, range(1, 11)))
pares
```
