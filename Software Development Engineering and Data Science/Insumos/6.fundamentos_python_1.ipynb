## 1. Números y Operaciones Aritméticas

Python maneja principalmente dos tipos numéricos: **enteros (`int`)** y **números de punto flotante (`float`)**.

* Ejemplo: `2` es un entero (`int`) y `2.0` es un decimal (`float`).

```python
# Suma
5 + 2
```

```python
# Resta
type(5 - 2.1)
```

```python
# Multiplicación
5 * 2
```

La división estándar se realiza con la barra inclinada `/`. En Python 3, la división siempre devuelve un número de punto flotante (`float`), incluso si ambos números son enteros.

```python
# División
5 / 2
```

Para calcular potencias (exponenciación), se utiliza el operador de doble asterisco `**`.

```python
# Exponente (5 al cuadrado)
5 ** 2
```

Python sigue la jerarquía estándar de operaciones matemáticas (PEMDAS/BODMAS). Por ejemplo, en `1 + 2 * 3 + 4`, primero se evalúa la multiplicación (`2 * 3`).

```python
# Orden de operaciones
1 + 2 * 3 + 4
```

**Buena práctica:** Es recomendable usar paréntesis `()` para definir explícitamente el orden de evaluación deseado.

```python
# Uso de paréntesis para controlar la precedencia
(1 + 2) * (3 + 4)
```

El operador **módulo (`%`)** calcula el **residuo de una división entera**.

* Es muy útil para determinar si un número es par o impar:
  * Si `numero % 2` retorna `0`, el número es **par**.
  * Si `numero % 2` retorna `1`, el número es **impar**.

```python
# Retorna 0: 6 es un número par
6 % 2
```

```python
# Retorna 1: 5 es un número impar (residuo de 5 / 2)
5 % 2
```

## 2. Variables y Asignación

Una variable permite almacenar datos en memoria asignándoles un nombre. La asignación se realiza mediante el operador `=`.<br>
Al llamar al nombre de la variable, obtenemos el valor asignado.

```python
# Asignación de valores a variables
x = 5
y = 3
```

```python
# Consultar el valor de x
x
```

```python
# Consultar el valor de y
y
```

Podemos operar directamente con las variables creadas:

```python
# Sumar x e y
x + y
```

```python
# Multiplicar x e y
x * y
```

También es posible almacenar el resultado de una operación en una nueva variable:

```python
# Restar y almacenar el resultado en 'z'
z = x - y
z
```

Las variables son dinámicas; podemos **reasignar** o sobrescribir su valor en cualquier momento:

```python
# Reasignar x con su valor al cuadrado
x = x * x
x
```

### Reglas y convenciones para nombrar variables:
* La convención estándar en Python es usar **snake_case** (minúsculas separadas por guiones bajos `_`), por ejemplo: `ingreso_total`, `tasa_conversion`, `nombre_cliente`.
* **No pueden** comenzar con números (ej. `1variable` generará error de sintaxis).
* **No pueden** contener caracteres especiales ni símbolos de puntuación (ej. `$variable`, `var!`).
* Son sensibles a mayúsculas y minúsculas (*case-sensitive*): `Variable` y `variable` son dos identificadores distintos.
* **No deben** coincidir con palabras reservadas del lenguaje (como `class`, `for`, `if`, `def`, etc.).

```python
nombre_de_variable = 20
nombre_de_variable
```

## 3. Cadenas de Texto (*Strings*)

Las cadenas de texto representan secuencias de caracteres y se definen utilizando comillas simples (`''`) o dobles (`""`).

```python
# Cadena con comillas simples
'Cadena de texto con comillas simples'
```

```python
# Cadena con comillas dobles
"Cadena de texto con comillas dobles"
```

```python
# Si el texto contiene comillas simples internas, delimitamos con comillas dobles
"Texto que incluye una comilla simple: 'ejemplo'"
```

Para imprimir valores formalmente en consola utilizamos la función `print()`:

```python
mensaje = "Fundamentos de Python"
print(mensaje)
```

### Formateo de Cadenas (*String Formatting*)

Existen formas estructuradas de insertar variables dentro de un texto:

1. **Método `.format()` posicional o por etiquetas.**
2. **f-Strings (Recomendado en Python moderno):** Anteponiendo una `f` antes de las comillas e insertando las variables directamente entre llaves `{}`.

```python
nombre = 'Carlos'
edad = 15

# Formato posicional con .format()
print('Mi nombre es {} y tengo {} años.'.format(nombre, edad))
```

```python
# Formato con nombres de parámetros en .format()
print('Usuario: {user}, Edad: {age}'.format(user=nombre, age=edad))
```

```python
# Uso moderno con f-Strings
print(f'Registro: {nombre} tiene {edad} años.')
```

### Indexación y Slicing en Cadenas

* Las cadenas son secuencias indexadas comenzando desde la posición `0`.
* El índice negativo `-1` accede al último elemento.
* Son **inmutables**: no se puede modificar un carácter individual por asignación directa (ej. `cadena[0] = 'A'` dará error).

```python
cadena = "Python"

# Primer elemento (índice 0)
cadena[0]
```

```python
# Último elemento (índice negativo)
cadena[-1]
```

```python
# Slicing: desde el índice 1 hasta el final
cadena[1:]
```

```python
# Slicing: desde el inicio hasta el índice 4 (exclusivo)
cadena[:4]
```

```python
cadena[2:4]
```

```python
fecha = '2026-08-24T12:30:00Z'
fecha[:10]
```

### Métodos Comunes de Cadenas
Los métodos más habituales para limpieza y transformación de texto en analítica son `.lower()`, `.upper()`, `.strip()` y `.split()`.

```python
texto = 'Analisis de Datos con Python'

# Convertir a minúsculas
print(texto.lower())

# Convertir a mayúsculas
print(texto.upper())
```

```python
# Dividir la cadena por espacios (retorna una lista de palabras)
texto.split()
```

```python
fecha.split('T')[0],fecha.split('T')[-1]
```

```python
# Dividir por un delimitador específico
etiqueta = 'dataset_ventas_2026.csv'
etiqueta.split('_')
```

## 4. Listas (*Lists*)

Las listas son colecciones ordenadas y **mutables** delimitadas por corchetes `[]`.

**Características principales:**
* Pueden almacenar cualquier combinación de tipos de datos.
* Son dinámicas (pueden crecer o reducirse de tamaño).
* Permiten indexación, reasignación y *slicing*.

```python
# Creación de listas
mi_lista = ['a', 'b', 'c']
mi_lista
```

```python
# Agregar un elemento al final con .append()
mi_lista.append('d')
mi_lista
```

```python
# Extraer y eliminar el último elemento con .pop()
elemento_removido = mi_lista.pop()
print('Elemento removido:', elemento_removido)
print('Lista resultante:', mi_lista)
```

```python
# Reasignación por índice (mutabilidad)
mi_lista[0] = 'inicio'
mi_lista
```

### Listas Anidadas (*Nested Lists*)
Podemos incluir listas dentro de otras listas, lo cual es la base estructural para matrices y datos multidimensionales.

```python
matriz = [1, 2, [3, 4, ['valor_interno']]]

# Acceso por niveles de profundidad
matriz[2][2][0]
```

## 5. Diccionarios (*Dictionaries*)

Los diccionarios son estructuras basadas en pares **clave-valor** (`key-value`) delimitadas por llaves `{}`.

* A diferencia de las listas, no se accede por posición numérica, sino a través de su **clave** (`key`).
* Son ampliamente utilizados para representar registros de datos y estructuras JSON.

```python
# Definición de un diccionario
metricas = {
    'usuarios': 1200,
    'tasa_rebote': 0.24,
    'fuentes': ['organico', 'directo', 'pago']
}

# Acceso por clave
metricas['usuarios']
```

```python
metricas['fuentes'][-1]
```

```python
# Acceso anidado a listas dentro del diccionario
metricas['fuentes'][0]
```

```python
# Métodos principales: keys(), values(), items()
print('Claves:', list(metricas.keys()))
print('Valores:', list(metricas.values()))
print('Pares clave-valor:', list(metricas.items()))
```

## 6. Tuplas (*Tuples*)

Las tuplas son secuencias ordenadas delimitadas por paréntesis `()`.

* Son **inmutables**: una vez creadas, no se pueden modificar, agregar ni reasignar elementos.
* Se utilizan para proteger la integridad de los datos frente a modificaciones accidentales.

```python
# Creación de una tupla
coordenadas = (10, 20, 30)
coordenadas[0]
```

```python
# Intentar modificar una tupla generará un TypeError:
coordenadas[0] = 50  # Descomentar para ver el error de inmutabilidad
```

## 7. Conjuntos (*Sets*)

Un conjunto es una colección no ordenada de **elementos únicos** definida con llaves `{}` o la función `set()`.

* Elimina automáticamente valores duplicados.
* Resulta muy útil para obtener categorías o valores únicos en un conjunto de datos.

```python
# Declaración con elementos repetidos
categorias = {1, 2, 2, 3, 3, 3, 4, 4, 4, 4}
categorias
```

```python
# Agregar elementos con .add()
categorias.add(5)
categorias
```

```python
categorias.add(5)
categorias
```

## 8. Booleanos y Operadores de Pertenencia

Los valores booleanos representan estados lógicos: `True` o `False` (iniciando siempre con mayúscula).
El operador `in` permite verificar de forma rápida la pertenencia de un elemento dentro de una colección.

```python
muestras = [10, 20, 30, 40]

# Comprobar si un elemento está presente
print(20 in muestras)
print(99 in muestras)
```

```python
for muestra in muestras:
    print(f'Valor de muestra: {muestra}')
```
