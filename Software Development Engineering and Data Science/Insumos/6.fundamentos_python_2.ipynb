## 1. Operadores de Comparación

Los operadores de comparación evalúan la relación entre dos operandos y devuelven un valor booleano (`True` o `False`).

| Operador | Significado |
| :--- | :--- |
| `>` | Mayor que |
| `<` | Menor que |
| `>=` | Mayor o igual que |
| `<=` | Menor o igual que |
| `==` | Igual a (comparación de valor) |
| `!=` | Diferente / Distinto de |

```python
# Comparación: Mayor que
3 > 4
```

```python
# Comparación: Menor que
3 < 4
```

```python
# Comparación: Mayor o igual que
2 >= 2
```

```python
# Comparación: Menor o igual que
3 <= 4
```

**Importante:** Recuerda que el operador `==` compara igualdad, mientras que un solo `=` es el operador de asignación de variables.

```python
# Igualdad de valores numéricos
1 == 1
```

También podemos comparar cadenas de texto. Las comparaciones distinguen mayúsculas de minúsculas (*case-sensitive*):

```python
# Distingue mayúsculas y minúsculas (retorna False)
'Datos' == 'DATOS'
```

```python
# Textos distintos
'Python' == 'SQL'
```

```python
# Cadenas exactamente idénticas
'Python' == 'Python'
```

```python
# Operador de desigualdad (!=)
'Python' != 'Python'
```

## 2. Operadores Lógicos

Permiten combinar múltiples expresiones condicionales:

* **`and`**: Retorna `True` **únicamente si todas las condiciones** evaluadas son verdaderas.
* **`or`**: Retorna `True` si **al menos una de las condiciones** es verdadera.
* **`not`**: Invierte el valor de verdad de una expresión.

```python
# Operador 'and': Ambas condiciones deben cumplirse
(1 < 2) and (2 < 3)
```

```python
# Operador 'or': Se cumple si al menos una es verdadera
(1 < 2) or (2 > 3)
```

```python
# Operador 'or': Ambas son falsas, retorna False
(1 > 2) or (2 > 3)
```

```python
# Encadenamiento múltiple con 'or'
(1 == 2) or (2 == 3) or (4 == 4)
```

## 3. Sentencias Condicionales (`if`, `elif`, `else`)

Las estructuras condicionales dirigen el flujo de ejecución de un programa evaluando condiciones lógicas.

### Estructura Básica (`if`)
El bloque dentro del `if` solo se ejecutará si la condición resulta `True`.

```python
# Condición verdadera: el bloque se ejecuta
if 5>3:
    print('La condición se cumple')
    
```

>&#9888;&#65039; **Regla clave de sintaxis: Indentación (Sangría)**<br>
>En Python los bloques de código no se delimitan con llaves `{}` como en C/Java/JS, sino mediante **espacios en blanco al inicio de la línea (generalmente 4 espacios o un tabulador)** después de los dos puntos `:`.

```python
# Condición falsa: el bloque NO se ejecuta y no muestra salida
if 3 < 2:
    print('La condición se cumple')
```

### Bifurcación Alternativa (`if - else`)
Cuando la condición del `if` es falsa, podemos especificar un camino alternativo mediante la cláusula `else`.

```python
valor = 10

if valor > 20:
    print('El valor supera el umbral de 20')
else:
    print('El valor es menor o igual a 20')
```

### Evaluación Múltiple (`if - elif - else`)
Para evaluar múltiples condiciones secuenciales utilizamos `elif` (*abreviatura de else if*).

**Regla fundamental de evaluación:**
* Python evalúa las condiciones de arriba hacia abajo.
* **Únicamente se ejecuta el primer bloque cuya condición sea `True`**. Una vez ejecutado, el resto de condiciones `elif` y `else` son ignoradas por completo.

```python
calificacion = 10

if calificacion >= 90:
    print('Rendimiento: Excelente')
elif calificacion >= 80:
    print('Rendimiento: Bueno')
elif calificacion >= 70:
    print('Rendimiento: Regular')
else:
    print('Rendimiento: Necesita mejorar')
```

### Demostración del cortocircuito en `elif`:
Observa cómo a pesar de que la tercera condición (`2 == 2`) también es verdadera, nunca llega a ejecutarse porque la segunda (`3 == 3`) se cumplió primero:

```python
if 1 == 2:
    print('Primera condición cumplida')
elif 3 == 3:
    print('Segunda condición cumplida (Se detiene aquí)')
elif 2 == 2:
    print('Tercera condición cumplida (Ignorada)')
else:
    print('Bloque final else')
```
